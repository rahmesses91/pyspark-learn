# Regex Characters — Quick Reference

| Character | Meaning | Example |
|-----------|---------|---------|
| `.` | Any character (except newline) | `a.c` → "abc", "a1c" |
| `\d` | Digit `[0-9]` | `\d+` → "123" |
| `\D` | Non-digit | `\D+` → "abc" |
| `\w` | Word char `[a-zA-Z0-9_]` | `\w+` → "hello_1" |
| `\W` | Non-word character | `\W` → " ", ".", "!" |
| `\s` | Whitespace (space, tab, newline) | `\s+` → "   " |
| `\S` | Non-whitespace | `\S+` → "word" |
| `*` | Zero or more | `a*` → "", "a", "aa" |
| `+` | One or more | `\d+` → "1", "123" |
| `?` | Zero or one | `colou?r` → "color", "colour" |
| `{n}` | Exactly n times | `\d{3}` → "123" |
| `{n,}` | n or more times | `\d{2,}` → "12", "123" |
| `{n,m}` | Between n and m times | `\d{2,4}` → "12", "1234" |
| `^` | Start of string | `^Hello` → "Hello world" |
| `$` | End of string | `world$` → "Hello world" |
| `\b` | Word boundary | `\bword\b` → whole word only |
| `\B` | Non-word boundary | `\Bword\B` → inside a word |
| `[]` | Character class | `[aeiou]` → any vowel |
| `[^]` | Negated class | `[^0-9]` → non-digit |
| `|` | Alternation (OR) | `cat|dog` → "cat" or "dog" |
| `()` | Capturing group | `(\d+)` → capture digits |
| `(?:)` | Non-capturing group | `(?:a|b)` → match but don't capture |
| `(?P<name>)` | Named group | `(?P<id>\d+)` → capture as "id" |
| `\` | Escape special char | `\.` → literal dot |
| `\n` | Newline | |
| `\t` | Tab | |
| `\r` | Carriage return | |

**Note:** Use raw strings `r"..."` in Python so backslashes are preserved.

# Python Regular Expressions (regex) — Basics

This notebook shows **what regex is** and **how to use it** in Python with the `re` module.

---
## 1. What is regex?

A **regular expression** is a pattern that describes *shapes* of text, not just literal strings.

- Literal: "error" matches only the letters e-r-r-o-r.
- Regex: `\d+` matches *one or more digits* — e.g. "0", "123", "999".

```python
import re

# Pattern: one or more digits
pattern = r"\d+"

print("Pattern:", repr(pattern))
print("Matches '123':", bool(re.search(pattern, "123")))
print("Matches 'abc':", bool(re.search(pattern, "abc")))
print("Matches 'Order 456':", bool(re.search(pattern, "Order 456")))
```

---
## 2. `re.match()` vs `re.search()`

| Function | Where it looks | Use when |
|----------|----------------|----------|
| `re.match(pattern, string)` | **Start** of string only | You expect the pattern at the beginning |
| `re.search(pattern, string)` | **Anywhere** in string | You want the first occurrence anywhere |

Both return a **match object** (or `None`). Use `.group()` to get the matched text.

```python
import re

s = "ERROR|2024-01-15|Connection timeout"

# match: pattern must be at the START
m1 = re.match(r"ERROR", s)
print("match(r'ERROR', s):", m1.group() if m1 else None)

m2 = re.match(r"\d+", s)  # digits at start? No.
print("match(r'\\d+', s):", m2)

# search: first occurrence ANYWHERE
m3 = re.search(r"\d{4}-\d{2}-\d{2}", s)
print("search(date pattern, s):", m3.group() if m3 else None)
```

```python
# Always check for None before calling .group()
def first_digits(text):
    m = re.search(r"\d+", text)
    return m.group() if m else None

print(first_digits("Order 12345 shipped"))  # 12345
print(first_digits("No numbers here"))       # None
```

---
## 3. `re.findall()` — get all matches

Returns a **list** of every non-overlapping match (or list of tuples if there are capturing groups).

```python
import re

text = "Prices: 10, 20, 30 and 40."
numbers = re.findall(r"\d+", text)
print(numbers)  # ['10', '20', '30', '40']
```

```python
# Email-like pattern: word@word.letters
text = "Contact support@example.com or sales@company.org"
emails = re.findall(r"\w+@\w+\.\w{2,4}", text)
print(emails)  # ['support@example.com', 'sales@company.org']
```

---
## 4. `re.sub()` — replace by pattern

**Syntax:** `re.sub(pattern, replacement, string)`  
Replaces every match with `replacement` (string or callable).

```python
import re

# Replace every run of non-alphanumeric chars with one underscore
s = "Hello,  world! How are you?"
normalized = re.sub(r"[^a-zA-Z0-9]+", "_", s)
print(normalized)  # Hello_world_How_are_you_
```

```python
# Mask digits with X
s = "Order 12345 and 67890"
masked = re.sub(r"\d", "X", s)
print(masked)  # Order XXXXX and XXXXX
```

---
## 5. Capturing groups

Use parentheses `( )` in the pattern to **capture** a part of the match. Access via `.group(1)`, `.group(2)` or `.groups()`.

```python
import re

text = "Order #12345 shipped on 2024-01-15"

order_match = re.search(r"#(\d+)", text)
date_match = re.search(r"(\d{4}-\d{2}-\d{2})", text)

order_id = order_match.group(1) if order_match else None
date_str = date_match.group(1) if date_match else None
print("Order ID:", order_id, "| Date:", date_str)
```

```python
# findall with one group returns list of that group's matches
text = "user_id=1001 action=click product_id=302"
keys = re.findall(r"(\w+)=", text)
values = re.findall(r"=(\w+)", text)
print("Keys:", keys)
print("Values:", values)
print("Dict:", dict(zip(keys, values)))
```

## 6. Common patterns quick reference

| Need | Pattern | Example |
|------|---------|--------|
| Digits | `\d+` | "123" |
| Word chars | `\w+` | "hello_1" |
| Date YYYY-MM-DD | `\d{4}-\d{2}-\d{2}` | "2024-01-15" |
| Optional part | `colou?r` | "color" or "colour" |
| One of | `(ERROR|WARN|INFO)` | log levels |
| Word boundary | `\bword\b` | whole word only |

Use **raw strings** (`r"..."`) so backslashes are not eaten by Python.

```python
import re

# Filter lines that start with ERROR or WARN
lines = ["INFO: started", "ERROR: failed", "WARN: retry", "INFO: done"]
bad = [s for s in lines if re.search(r"^(ERROR|WARN):", s)]
print(bad)

# Extract the level from each matching line
levels = [re.search(r"^(ERROR|WARN):", s).group(1) for s in lines if re.search(r"^(ERROR|WARN):", s)]
print(levels)
```

| Goal | Function | Returns |

| Match from start | `re.match(pattern, s)` | Match or None |
| First match anywhere | `re.search(pattern, s)` | Match or None |
| All matches | `re.findall(pattern, s)` | List of strings (or tuples) |
| Replace matches | `re.sub(pattern, repl, s)` | New string |
| Split by pattern | `re.split(pattern, s)` | List of substrings |
